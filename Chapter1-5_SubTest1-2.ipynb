## **ข้อ 2** ให้สร้าง Web Application


**คำชี้แจง**
ให้นักศึกษาพัฒนา Web Application เพื่อประมวลผลข้อความ (ไทย/อังกฤษ) โดยให้นักศึกษาเลือก 1 หัวข้อที่สนใจ หรือจะคิดขึ้นใหม่ก็ได้

นักศึกษาสามารถใช้ AI เป็นผู้ช่วยในการเขียนโปรแกรมในครั้งนี้ได้

**เงื่อนไข**

นักศึกษาจะต้อง นำโค้ดทั้งหมดขึ้น GitHub Repository ส่วนตัว และทำการ Deploy แอปพลิเคชันผ่าน Streamlit Community Cloud เพื่อให้สามารถเข้าใช้งานและแสดงผลผ่าน URL หน้าเว็บจริงได้

**ตัวอย่างแนวทางหัวข้อ**

1. ระบบคัดกรองรีวิวสินค้า/อาหาร --> สกัดชื่อแบรนด์, พิกัดร้าน, เมนู, คำชม/คำติ
2. ระบบวิเคราะห์โพสต์เตือนภัย/ข่าวอุบัติเหตุ --> สกัดสถานที่เกิดเหตุ, วันเวลา, ผู้บาดเจ็บ, และหน่วยงานช่วยเหลือ
3. ระบบสกัดข้อมูลประกาศรับสมัครงาน --> สกัดตำแหน่งงาน, ชื่อบริษัท, เงินเดือน, สกิลที่ต้องการ, และสถานที่
4. ระบบวิเคราะห์ข้อความจองทริปท่องเที่ยว --> สกัดสถานที่เที่ยว, โรงแรม, จำนวนวัน, และงบประมาณ
5. ระบบกรองข้อความร้องเรียนบริการสาธารณะ --> สกัดประเภทปัญหา, พิกัดที่พบ, ลบข้อมูลส่วนตัว, และระดับความเร่งด่วน

**ข้อกำหนดทางเทคนิค**

ให้เลือกนำเทคนิคจากที่เรียนมา มาประยุกต์ใช้ ได้แก่

- Regex & Cleansing --> ลบ Noise หรือข้อมูลสำคัญ เช่น เบอร์โทร, ลิงก์

- Tokenization & Normalization --> ตัดคำ ลบ Stopwords หรือจัดการคำพิมพ์ผิด/คำลากเสียง

- Topic Identification --> จัดกลุ่มหัวข้อหลักของข้อความ (เช่น จำแนกประเภทงาน, ปัญหา, หรือสินค้า)

- POS & NER --> สกัด Named Entities (คน, สถานที่, เวลา, องค์กร) หรือใช้ POS Tagging ดึงคำสำคัญ

**สิ่งที่ต้องส่ง**
1. URL หน้าเว็บ Streamlit --> ลิงก์แอปพลิเคชันที่ Deploy บน Streamlit Community Cloud ที่เปิดใช้งานได้จริง

2. GitHub Repository URL --> ลิงก์ซอร์สโค้ดระบบ ซึ่งต้องมีโครงสร้างไฟล์ครบถ้วน ดังนี้:

    - app.py (ไฟล์หลักของ Streamlit App)

    - requirements.txt (ระบุไลบรารีที่ต้องใช้ เช่น streamlit, spacy, pythainlp, pandas)

    - README.md (อธิบายวิธีใช้งาน, แนวคิดของ Domain และตัวอย่าง Prompt ที่ใช้สั่ง AI)

    - ไฟล์ข้อมูลที่ใช้ในการทดสอบระบบ

**เกณฑ์การให้คะแนน**

1. ไฟล์ข้อมูลที่ใช้ในการทดสอบ (1 คะแนน)

2. นำโค้ดขึ้น GitHub และหน้าเว็บ Streamlit สามารถเข้าใช้งานได้จริง (2 คะแนน)

3. ความคิดสร้างสรรค์ในการเลือกหัวข้อ และการออกแบบสกัดข้อมูลที่สอดคล้องกับหัวข้อ (3 คะแนน)

4. การนำเทคนิค NLP จากที่เรียนมาประยุกต์ใช้ได้อย่างถูกต้องและสอดคล้องกับหัวข้อ (3 คะแนน)

5. หน้าเว็บใช้งานง่าย (1 คะแนน)
